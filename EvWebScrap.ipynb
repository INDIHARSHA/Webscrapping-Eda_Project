```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = (r"https://ev-database.org/#group=vehicle-group&rs-pr=10000_100000&rs-er=0_1000&rs-ld=0_1000&rs-ac=2_23&rs-dcfc=0_300&rs-ub=10_200&rs-tw=0_2500&rs-ef=100_350&rs-sa=-1_5&rs-w=1000_3500&rs-c=0_5000&rs-y=2010_2030&s=1&p=0-50")
response=requests.get(url)
```

```python
response
```

```python
soup = BeautifulSoup(response.text, "html.parser")
```

```python
brand=[]
model=[]
price=[]
mileage=[]
battery_capacity=[]
weight=[]
```

```python
brands=soup.find_all("a", class_="title")
models=soup.find_all("span",class_="model")
prices=soup.find_all("span",class_="country_uk")
mileages=soup.find_all("span",class_="erange_real")
batteries=soup.find_all("span",class_="battery_p")
weights=soup.find_all("span",class_="weight_p")

for b in brands:
    name=b.text.strip().split("\n")[0].split()[0]
    brand.append(name)
for m in models:
    name=m.text.strip()
    model.append(name)
for p in prices:
    name=p.text.strip()
    price.append(name)
for M in mileages:
    name=M.text.strip()
    mileage.append(name)
for B in batteries:
    name=B.text.strip()
    battery_capacity.append(name)
for w in weights:
    name=w.text.strip()
    weight.append(name)
```

```python
data = {
    "Brand": brand,
    "Model": model,
    "Price": price,
    "Range": mileage,
    "Battery": battery_capacity,
    "Weight": weight,
}
```

```python
ev=pd.DataFrame(data)
```
